```python

# Inputs
exam = 30
attendance = 30
study_hours = 4

# Weights
w1 = 0.5
w2 = 0.3
w3 = 0.2

# Bias
bias = -50

# Weighted Sum (z)
z = (exam * w1 + attendance * w2 + study_hours * w3 + bias)
print("Weighted Sum (z):", z) # Outputs: 20.3

# Step Activation Function
prediction = 1 if z >= 0 else 0

if prediction == 1:
    print("Student Passed") # Executed!
else:
    print("Student Failed")
```
